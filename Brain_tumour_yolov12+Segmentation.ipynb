```python
from ultralytics import YOLO
from ultralytics import SAM
from IPython.display import Image, display

# Load your pretrained YOLO brain tumor detector
yolo_model = YOLO("/content/runs/detect/train/weights/best.pt")

# Run inference on the test MRI image
results =
yolo_model("/content/Labeled-MRI-Brain-Tumor-Dataset-1/test/images/Tr-gl_0016_jpg.rf.99746694ea97fe0b73108832b462d48e.jpg")

# Load SAM 2.1 base model for segmentation
sam_model = SAM("sam2.1_b.pt")

# Iterate over YOLO results and segment detected tumors with SAM
for result in results:
    class_ids = result.boxes.cls.int().tolist()
    if len(class_ids) > 0:
        boxes = result.boxes.xyxy  # get bounding boxes in [x1, y1, x2, y2] format
        # Run SAM segmentation on original image and YOLO boxes
        sam_results = sam_model(result.orig_img, bboxes=boxes, verbose=False, save=True, device=0)
```
